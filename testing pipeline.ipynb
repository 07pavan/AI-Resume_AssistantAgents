```python
def read_resume(file_path):
    text = ""
    doc = fitz.open(file_path)

    for page in doc:
        text += page.get_text()

    return text
```

```python
resume_text = read_resume("resume.pdf")
resume_text = " ".join(resume_text.split())
resume_text = resume_text[:1800]
```

Create Agents

```python
from crewai import LLM
import os

# Groq LLM
groq_llm = LLM(
    model="groq/llama-3.1-8b-instant",
    temperature=0.2,
    max_tokens=800
)
openrouter_llm = LLM(
    model="openrouter/meta-llama/llama-3.1-8b-instruct",
    temperature=0.3,
    max_tokens=800
)
```

**Agents**

```python
resume_analyzer = Agent(
    role="Resume Analyzer",
    goal="Analyze resume and give feedback",
    backstory="Expert HR reviewer",
    llm=groq_llm,
    verbose=True
)
```

```python
resume_improver = Agent(
    role="Resume Writer",
    goal="Improve resume professionally",
    backstory="Professional resume writer",
    llm=openrouter_llm,
    verbose=True
)
```

```python
job_researcher = Agent(
    role="Job Researcher",
    goal="Find relevant jobs",
    backstory="Expert recruiter",
    llm=groq_llm,
    verbose=True
)
```

```python
cover_letter_agent = Agent(
    role="Cover Letter Writer",
    goal="Write cover letters",
    backstory="Career consultant",
    llm=openrouter_llm,
    verbose=True
)
```

**Tasks**

```python
analyze_resume_task = Task(
    description=f"""
    Analyze the following resume and give feedback.

    Resume:
    {resume_text}
    """,
    expected_output="A detailed analysis of the resume including strengths, weaknesses, and improvement suggestions.",
    agent=resume_analyzer
)
```

```python
improve_resume_task = Task(
    description=f"""
    Improve this resume and make it more professional.

    Resume:
    {resume_text}
    """,
    expected_output="A rewritten and improved professional resume with better formatting, stronger wording, and clearer achievements.",
    agent=resume_improver
)
```

```python
job_search_task = Task(
    description=f"""
    Based on the candidate resume below, find 5 relevant job opportunities.

    Resume:
    {resume_text}

    Include:
    - Job title
    - Company
    - Location
    """,
    expected_output="A list of 5 relevant job opportunities including title, company and location.",
    agent=job_researcher
)
```

```python
cover_letter_task = Task(
    description=f"""
    Write a professional cover letter using the following resume.

    Resume:
    {resume_text}
    """,
    expected_output="A professional and personalized cover letter suitable for job applications.",
    agent=cover_letter_agent
)
```

**Create Crew**

```python
crew = Crew(
    agents=[
        resume_analyzer,
        resume_improver,
        job_researcher,
        cover_letter_agent
    ],
    tasks=[
        analyze_resume_task,
        improve_resume_task,
        job_search_task,
        cover_letter_task
    ],
    verbose=True
)
```

**run**

```python
import time

result = crew.kickoff()
time.sleep(25)

print(result)
```
